2. Check Palindrome String


Interview Question: What is a palindrome? How can you check it in Python?

Coding Prompt: Write a program to check whether a given string reads the same forward and backward.

```python
text = "madam"

if text == text[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

3. Find Factorial of a number.

Interview Question: What is factorial? How can you calculate it using Python?

Coding Prompt: Write a program to calculate the factorial of a given positive integer using a loop.

```python
num = 5
fact = 1

for i in range(1, num + 1):
    fact *= i

print(fact)
```

4. Generate Fibonacci Series

Interview Question: What is the Fibonacci sequence?

Coding Prompt: Write a Python program to generate the first 10 numbers in the Fibonacci series.

```python
a, b = 0, 1

for i in range(10):
    print(a, end=" ")
    a, b = b, a + b
```

5. Check Prime number

Interview Question: How do you identify a prime number?

Coding Prompt: Write a Python program to check whether a given number is prime or not.

```python
num = 17

if num < 2:
    print("Not Prime")
else:
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
```

# Section 2: List and Collections

1.Find Largest element in List

Interview Question: How do you find the largest element without using max()?

Coding Prompt: Write a program to find the largest number in a list using a loop.

```python
numbers = [10, 25, 5, 40, 15]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print(largest)
```

2. Find Largest number

Interview Question: How do you find the second-largest distinct element in a list?

Coding Prompt: Write a Python program to find the second-largest distinct number without sorting the list.

```python
numbers = [10, 20, 40, 30, 40]

unique = list(set(numbers))
unique.remove(max(unique))

print(max(unique))
```

3. Remove Duplicates from a List


Interview Question: How do you remove duplicate elements from a list?

Coding Prompt: Write a program to remove duplicates while maintaining the original order.

```python
numbers = [1, 2, 2, 3, 4, 4, 5]

result = []

for num in numbers:
    if num not in result:
        result.append(num)

print(result)
```

4. Count frequency of elements

Interview Question: How do you count the frequency of elements in a list?

Coding Prompt: Write a program to count how many times each element occurs.

```python
numbers = [1, 2, 2, 3, 3, 3]

frequency = {}

for num in numbers:
    frequency[num] = frequency.get(num, 0) + 1

print(frequency)
```

5. Find common elements in two lists

Interview Question: How do you find common elements between two lists?

Coding Prompt: Write a program to identify elements that are present in both lists.

```python
a = [1, 2, 3, 4]
b = [3, 4, 5, 6]

common = list(set(a) & set(b))

print(common)
```

# Section 3: Strings

1. Count vowels in a string

Interview Question: How do you count vowels in a string?

Coding Prompt: Write a program to count the number of vowels in a given string.

```python
text = "Programming"
vowels = "aeiou"

count = 0

for ch in text.lower():
    if ch in vowels:
        count += 1

print(count)
```

2. Find first non repeating character

Interview Question: How do you find the first character that appears only once?

Coding Prompt: Write a program to find the first non-repeating character in a string.

```python
text = "aabbcde"

for ch in text:
    if text.count(ch) == 1:
        print(ch)
        break
```

## Section 4: Functions and OOP

1.Check Even or Odd Using a Function

Interview Question: How do you define and call a function in Python?

Coding Prompt: Create a function that accepts a number and returns whether it is even or odd.

```python
def check_number(num):
    if num % 2 == 0:
        return "Even"
    return "Odd"

print(check_number(10))
```

2. Demonstrate Inheritance

Interview Question: What is inheritance in Python?

Coding Prompt: Create a parent class named Animal and a child class named Dog that inherits its method.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

d = Dog()
d.sound()
d.bark()
```

3. Demonstrate Encapsulation

Interview Question: What is encapsulation in Python?

Coding Prompt: Create a class with a private attribute and access it using a getter method.

```python
class Student:
    def __init__(self, name):
        self.__name = name

    def get_name(self):
        return self.__name

s = Student("Kavya")
print(s.get_name())
```

4. Generator Function

Interview Question: What is a generator in Python?

Coding Prompt: Create a generator that yields numbers from 1 to 5 using the yield keyword.

```python
def generate_numbers():
    for i in range(1, 6):
        yield i

for num in generate_numbers():
    print(num)
```
