# CREATE TABLES

```python
%%sql

CREATE TABLE IF NOT EXISTS Department (
    dept_id INT PRIMARY KEY,
    dept_name VARCHAR(50) NOT NULL,
    location VARCHAR(50)
);

CREATE TABLE IF NOT EXISTS Employee (
    emp_id INT PRIMARY KEY,
    emp_name VARCHAR(100) NOT NULL,
    salary DECIMAL(10,2),
    dept_id INT,
    manager_id INT,
    hire_date DATE,
    FOREIGN KEY (dept_id) REFERENCES Department(dept_id)
);

CREATE TABLE IF NOT EXISTS Customers (
    customer_id INT PRIMARY KEY,
    customer_name VARCHAR(100),
    city VARCHAR(50),
    email VARCHAR(100)
);

CREATE TABLE IF NOT EXISTS Orders (
    order_id INT PRIMARY KEY,
    customer_id INT,
    order_date DATE,
    amount DECIMAL(10,2),
    status VARCHAR(30),
    FOREIGN KEY (customer_id) REFERENCES Customers(customer_id)
);

CREATE TABLE IF NOT EXISTS Products (
    product_id INT PRIMARY KEY,
    product_name VARCHAR(100),
    category VARCHAR(50),
    price DECIMAL(10,2)
);

CREATE TABLE IF NOT EXISTS OrderItems (
    item_id INT PRIMARY KEY,
    order_id INT,
    product_id INT,
    quantity INT,
    FOREIGN KEY (order_id) REFERENCES Orders(order_id),
    FOREIGN KEY (product_id) REFERENCES Products(product_id)
);

CREATE TABLE IF NOT EXISTS Course (
    course_id INT PRIMARY KEY,
    course_name VARCHAR(100),
    duration_months INT
);

CREATE TABLE IF NOT EXISTS Student (
    student_id INT PRIMARY KEY,
    student_name VARCHAR(100),
    marks DECIMAL(5,2),
    course_id INT,
    FOREIGN KEY (course_id) REFERENCES Course(course_id)
);
```

Q1. Find the second-highest salary.

Interview Question: How do you find the second-highest salary from an Employee table?

Coding Prompt: Write a MySQL query to retrieve the second-highest distinct salary from the Employee table.

```python
%%sql

SELECT MAX(salary) AS second_highest_salary
FROM Employee
WHERE salary < (
    SELECT MAX(salary)
    FROM Employee
);
```

Q2. Find employees earning more than the average salary.

Interview Question: How do you retrieve employees whose salary is greater than the company's average salary?

Coding Prompt: Write a query to display employee names and salaries above the overall average salary.

```python
%%sql

SELECT emp_name, salary
FROM Employee
WHERE salary > (
    SELECT AVG(salary)
    FROM Employee
);
```

Q3. Find duplicate salaries.

Interview Question: How do you identify duplicate values in a table?

Coding Prompt: Write a query to find salaries that are shared by more than one employee.

```python
%%sql

SELECT salary, COUNT(*) AS employee_count
FROM Employee
GROUP BY salary
HAVING COUNT(*) > 1;
```

Q4. Find the number of employees in each department.

Interview Question: How do you count employees department-wise?

Coding Prompt: Display department IDs and the number of employees in each department.

```python
%%sql

SELECT
    dept_id,
    COUNT(*) AS total_employees
FROM Employee
GROUP BY dept_id;
```

Q5. Display employee names with department names.

Interview Question: How do you retrieve data from two related tables?

Q7. Find customers who have placed more than one order.

Interview Question: How do you identify customers who have placed multiple orders?

Coding Prompt: Display customer names and their total order counts for customers having more than one order.

```python
%%sql

SELECT
    c.customer_name,
    COUNT(o.order_id) AS total_orders
FROM Customers c
INNER JOIN Orders o
ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.customer_name
HAVING COUNT(o.order_id) > 1;
```

Q8. Find the highest-paid employee in each department.

Interview Question: How do you find the highest salary department-wise?

Coding Prompt: Retrieve employee names, salaries, and department IDs for employees earning the maximum salary in each department.

```python
%%sql

SELECT
    emp_name,
    salary,
    dept_id
FROM Employee e
WHERE salary = (
    SELECT MAX(salary)
    FROM Employee
    WHERE dept_id = e.dept_id
);
```

Q9. Find products that have never been ordered.

Interview Question: How do you find records that do not have matching records in another table?

Coding Prompt: Retrieve products that do not appear in the OrderItems table.

```python
%%sql

SELECT
    p.product_name
FROM Products p
LEFT JOIN OrderItems oi
ON p.product_id = oi.product_id
WHERE oi.product_id IS NULL;
```

Q10. Rank employees based on salary.

Interview Question: How do you rank employees according to their salaries?

Coding Prompt: Assign a salary rank to each employee, with the highest salary receiving rank 1.

Answer:

```python
%%sql

SELECT
    emp_name,
    salary,
    DENSE_RANK() OVER (
        ORDER BY salary DESC
    ) AS salary_rank
FROM Employee;
```

Level 2 – Aggregate Functions and Joins

Q16. Calculate the total salary expenditure for each department.

Interview Question: How do you calculate department-wise total salary?

Coding Prompt: Display department names and the sum of salaries paid to employees in each department.

```python
%%sql

SELECT
    d.dept_name,
    SUM(e.salary) AS total_salary
FROM Department d
INNER JOIN Employee e
    ON d.dept_id = e.dept_id
GROUP BY d.dept_id, d.dept_name;
```

Q17. Find the total quantity ordered for each product.

Interview Question: How do you calculate the total quantity sold for each product?

Coding Prompt: Display product names and the total quantities ordered.

```python
%%sql

SELECT
    p.product_name,
    SUM(oi.quantity) AS total_quantity
FROM Products p
INNER JOIN OrderItems oi
    ON p.product_id = oi.product_id
GROUP BY p.product_id, p.product_name;
```

Q18. Find the course with the highest average student marks.

Interview Question: How do you find the group with the highest average value?

Coding Prompt: Retrieve the course name and average marks of the course with the highest average student marks.

```python
%%sql

SELECT
    c.course_name,
    AVG(s.marks) AS average_marks
FROM Course c
INNER JOIN Student s
    ON c.course_id = s.course_id
GROUP BY c.course_id, c.course_name
ORDER BY average_marks DESC
LIMIT 1;
```

Q19. Find the department with the highest average salary.

Interview Question: How do you identify the department that pays the highest average salary?

Coding Prompt: Retrieve the department name and its average employee salary.

```python
%%sql

SELECT
    d.dept_name,
    AVG(e.salary) AS average_salary
FROM Department d
INNER JOIN Employee e
    ON d.dept_id = e.dept_id
GROUP BY d.dept_id, d.dept_name
ORDER BY average_salary DESC
LIMIT 1;
```

Q20. Find customers whose total order amount exceeds 5,000.

Interview Question: How do you filter customers based on their total spending?

Coding Prompt: Display customer names and total order amounts for customers whose combined order amount exceeds 5,000.

```python
%%sql

SELECT
    c.customer_name,
    SUM(o.amount) AS total_spending
FROM Customers c
INNER JOIN Orders o
    ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.customer_name
HAVING SUM(o.amount) > 5000;
```

Level 3 – Subqueries and Self Joins

Q21. Find employees who earn more than their managers.

Interview Question: How do you compare records within the same table?

Coding Prompt: Retrieve employees whose salary is greater than their manager's salary using a self join.

```python
%%sql

SELECT
    e.emp_name AS employee_name,
    e.salary AS employee_salary,
    m.emp_name AS manager_name,
    m.salary AS manager_salary
FROM Employee e
INNER JOIN Employee m
    ON e.manager_id = m.emp_id
WHERE e.salary > m.salary;
```

Q22. Find the third-highest distinct salary.

Interview Question: How do you find the third-highest salary without using LIMIT?

Coding Prompt: Retrieve the third-highest distinct salary using a correlated subquery.

```python
%%sql

SELECT DISTINCT salary
FROM Employee e1
WHERE 2 = (
    SELECT COUNT(DISTINCT salary)
    FROM Employee e2
    WHERE e2.salary > e1.salary
);
```

Q23. Find employees who earn the same salary as another employee.

Interview Question: How do you identify employees with matching salary values?

Coding Prompt: Display employee names and salaries where at least one other employee has the same salary.

```python
%%sql

SELECT
    e1.emp_name,
    e1.salary
FROM Employee e1
WHERE EXISTS (
    SELECT 1
    FROM Employee e2
    WHERE e1.salary = e2.salary
      AND e1.emp_id <> e2.emp_id
);
```
